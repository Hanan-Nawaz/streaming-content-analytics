7. We notice, we have columns where data is either 0 or an empty list. 
    - We are checking columns to find the number of these zero or empty list values.

```python
cols_to_check_empty_list = ["genres", "production_countries", "spoken_languages"]
cols_int = ["revenue", "runtime"]
cols_float = ['budget', 'popularity']
cols_str = df.select_dtypes(object).columns.tolist()
```

```python
for col in cols_to_check_empty_list:
    print(f"Values that contain emply list in {col}: {df[df[col] == "[]"].shape[0]}")

for col in cols_str:
    print(f"Values that contain emply string in {col}: {df[df[col] == ""].shape[0]}")

for col in cols_int + cols_float:
    print(f"Values that contain 0 in {col}: {df[df[col] == 0].shape[0]}")
```

8. Replacing null values and zero's with NA, so we could drop them.

```python
for col in cols_to_check_empty_list:
    df[col] = df[col].replace("[]", pd.NA)

for col in cols_int + cols_float:
    df[col] = df[col].replace(0, pd.NA)
```

```python
for col in cols_to_check_empty_list:
    print(f"Values that contain emply list in {col}: {df[df[col] == "[]"].shape[0]}")

for col in cols_str:
    print(f"Values that contain emply string in {col}: {df[df[col] == ""].shape[0]}")

for col in cols_int + cols_float:
    print(f"Values that contain 0 in {col}: {df[df[col] == 0].shape[0]}")
```

10. using Json to convert ["genres", "production_countries", "spoken_languages"] into python list.
    - Before: '[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]'
    - After: [Action, Adventure, Fantasy, Science Fiction]

```python
def parse_string(string):
    try:
        return [d["name"] for d in json.loads(string)]
    except Exception as e:
        print(f"Error is {e}")
```

```python
df["genres"] = df["genres"].apply(parse_string)
df["production_countries"] = df["production_countries"].apply(parse_string)
df["spoken_languages"] = df["spoken_languages"].apply(parse_string)
```

11. Converting the datatypes of columns to those types where they better fit and reseting indexes.

```python
df["release_date"] = pd.to_datetime(df["release_date"])

convert_to_int = ["runtime", "vote_count", "revenue"]
convert_to_Int64 = ["budget"]
convert_to_float = ["popularity", "vote_average"]

df[convert_to_int] = df[convert_to_int].astype(np.int64)
df[convert_to_Int64] = df[convert_to_Int64].astype("Int64")
df[convert_to_float] = df[convert_to_float].astype(np.float64)

df.reset_index(drop=True, inplace=True)
```

12. Rounding the popularty to decimal point of 2 for better and consistent analysis.


```python
df["popularity"] = df["popularity"].round(2)
```
